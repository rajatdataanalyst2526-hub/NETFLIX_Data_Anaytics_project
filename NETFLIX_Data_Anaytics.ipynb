### Language Analysis:

```python
# Create Language Mapping from Country
language_map = {
    'United States': 'English',
    'United Kingdom': 'English',
    'Canada': 'English',
    'India': 'Hindi',
    'Japan': 'Japanese',
    'South Korea': 'Korean',
    'France': 'French',
    'Spain': 'Spanish',
    'Mexico': 'Spanish',
    'Germany': 'German',
    'Italy': 'Italian',
    'China': 'Chinese',
    'Turkey': 'Turkish'
}
```

```python
# Extract first country if multiple countries exist
df['main_country'] = df['country'].str.split(',').str[0].str.strip()

# Map language
df['language'] = df['main_country'].map(language_map)

# Replace missing mapped values
df['language'] = df['language'].fillna('Other')
```

```python
# Language Distribution

print("Language Distribution:")
print(df['language'].value_counts())


# Visualization 1: Countplot

plt.figure(figsize=(10,6))
sns.countplot(y='language', data=df,
              order=df['language'].value_counts().index,
              palette='Set2')

plt.title("Distribution of Content by Language")
plt.xlabel("Count")
plt.ylabel("Language")
plt.show()
```

```python
# Visualization 2: Pie Chart

lang_count = df['language'].value_counts()

plt.figure(figsize=(8,8))
plt.pie(lang_count.values,
        labels=lang_count.index,
        autopct='%1.1f%%',
        startangle=140)

plt.title("Language Share of Netflix Content")
plt.show()
```

```python
# Visualization 3: Language by Type

plt.figure(figsize=(10,6))
sns.countplot(data=df, y='language', hue='type',
              order=df['language'].value_counts().index,
              palette='coolwarm')

plt.title("Language Distribution by Content Type")
plt.xlabel("Count")
plt.ylabel("Language")
plt.show()
```

### Content Evolution Over Time:

```python
# Convert Duration into Numeric
# Example: "90 min" -> 90 , "2 Seasons" -> 2
df['duration_num'] = df['duration'].str.extract('(\d+)').astype(float)
```

```python
# Convert Rating into Numeric Codes

df['rating_code'] = df['rating'].astype('category').cat.codes
```

```python
# Group by Release Year

yearly_data = df.groupby('release_year').agg({
    'duration_num': 'mean',
    'rating_code': 'mean',
    'show_id': 'count'
}).reset_index()

yearly_data.rename(columns={'show_id': 'content_count'}, inplace=True)

print(yearly_data.head())
```

```python
# Visualization 1: Average Duration Over Time
plt.figure(figsize=(12,5))
sns.lineplot(x='release_year', y='duration_num', data=yearly_data, marker='o')
plt.title("Average Duration Over Years")
plt.xlabel("Release Year")
plt.ylabel("Average Duration")
plt.show()
```

```python
# Visualization 2: Average Rating Trend
plt.figure(figsize=(12,5))
sns.lineplot(x='release_year', y='rating_code', data=yearly_data, marker='o', color='green')
plt.title("Rating Trend Over Years")
plt.xlabel("Release Year")
plt.ylabel("Average Rating Code")
plt.show()
```

```python
# Visualization 3: Content Growth Over Time
plt.figure(figsize=(12,5))
sns.barplot(x='release_year', y='content_count', data=yearly_data, palette='viridis')

plt.title("Number of Titles Released Each Year")
plt.xlabel("Release Year")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

```python
# Visualization 4: Duration by Type
plt.figure(figsize=(10,5))
sns.boxplot(x='type', y='duration_num', data=df, palette='Set2')

plt.title("Duration Distribution by Content Type")
plt.xlabel("Type")
plt.ylabel("Duration")
plt.show()
```

```python
# Summary
print("Average Duration Increased/Decreased over years can be seen above.")
print("Content production trend also visible year-wise.")
```
